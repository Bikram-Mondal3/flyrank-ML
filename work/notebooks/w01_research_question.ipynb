## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

The decision is: which content items should be prioritized for potential refresh or review? The person acting on the output would be an SEO or content team member. Their action would be to manually review the prioritized content and, when the evidence supports it, potentially refresh, expand, protect, prune, or monitor that content.

This is a decision-support problem, not simply a plan to train a model. A false positive could waste team time and resources on content that did not need a refresh. A false negative could miss content that may have had a useful improvement opportunity. Data can help because the team is choosing among many content items with different levels of demand, visibility, and click activity.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one ? typing sentences here breaks Run All.
from pathlib import Path
import pandas as pd

candidate_paths = [
    Path("data/raw/content_refresh_anonymized.csv"),
    Path("../../data/raw/content_refresh_anonymized.csv"),
]

csv_path = next((path for path in candidate_paths if path.exists()), None)
if csv_path is None:
    raise FileNotFoundError("Could not find data/raw/content_refresh_anonymized.csv")

df = pd.read_csv(csv_path)

print("Rows:", len(df))
print("Columns:", df.columns.tolist())
print(f"Total content items: {len(df):,}")

if "search_volume" in df.columns:
    print(f"Median search volume: {df['search_volume'].median():,.0f}")

if "impressions_90d" in df.columns:
    print(f"Median 90-day impressions: {df['impressions_90d'].median():,.0f}")

if "clicks_90d" in df.columns:
    print(f"Median 90-day clicks: {df['clicks_90d'].median():,.0f}")
```

```python
summary_columns = [
    col for col in ["search_volume", "impressions_90d", "clicks_90d"]
    if col in df.columns
]

df[summary_columns].describe().round(2)
```

```python
if {"search_volume", "impressions_90d"}.issubset(df.columns):
    corr = df["search_volume"].corr(df["impressions_90d"])
    print(
        "Observed correlation between search volume and 90-day impressions: "
        f"{corr:.3f}"
    )
```

```python
available_refresh_signals = [
    col for col in ["search_volume", "impressions_90d", "clicks_90d"]
    if col in df.columns
]

print("Signals I can use for initial refresh framing:", available_refresh_signals)
print("position_90d present?", "position_90d" in df.columns)
```
